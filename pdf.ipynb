```python
# A helper function to escape special PDF characters.
def escape_pdf_string(text):
    return text.replace('\\', '\\\\').replace('(', '\\(').replace(')', '\\)')

data = {
    "title": "My Title",
    "content": "This is the body text.\nIt has multiple lines.\nAnd even (parentheses) and backslashes \\ must work!"
}

# Escape the text before using it in the PDF stream
title_text = escape_pdf_string(data["title"])
body_text_lines = [escape_pdf_string(line) for line in data["content"].split('\n')]

pdf_objects = []

# Object 1: Catalog
pdf_objects.append("1 0 obj\n<< /Type /Catalog /Pages 2 0 R >>\nendobj")

# Object 2: Page Tree
pdf_objects.append("2 0 obj\n<< /Type /Pages /Kids [3 0 R] /Count 1 >>\nendobj")

# Object 3: Page Object (with reference to Font object)
pdf_objects.append("3 0 obj\n<< /Type /Page /Parent 2 0 R /MediaBox [0 0 595 842] /Contents 4 0 R /Resources << /Font << /F1 5 0 R >> >> >>\nendobj")

# --- CHANGE IS HERE: Setting the Leading (Line Spacing) ---

stream_lines = []
stream_lines.append("BT")                  # Begin Text object
stream_lines.append("/F1 24 Tf")           # Set Font to /F1, size 24
stream_lines.append("28 TL")               # <-- NEW: Set leading to 28 for the 24pt font
stream_lines.append("100 780 Td")          # Position the cursor (x, y)
stream_lines.append(f"({title_text}) Tj")  # Show the title text
stream_lines.append("T*")                  # Move to the next line (now moves down by 28 points)

stream_lines.append("/F1 12 Tf")           # Change Font to /F1, size 12
stream_lines.append("15 TL")               # <-- NEW: Set leading to 15 for the 12pt font
stream_lines.append("0 -15 Td")            # Nudge the cursor down a little extra for a gap

# Loop through the body lines and create PDF instructions for each
for line in body_text_lines:
    # We now only need Tj and T* because the leading is correctly set
    stream_lines.append(f"({line}) Tj")
    stream_lines.append("T*")

stream_lines.append("ET")                  # End Text object
stream_data = "\n".join(stream_lines)

# Object 4: The Content Stream
pdf_objects.append(f"4 0 obj\n<< /Length {len(stream_data)} >>\nstream\n{stream_data}\nendstream\nendobj")

# Object 5: The Font Definition
pdf_objects.append("5 0 obj\n<< /Type /Font /Subtype /Type1 /Name /F1 /BaseFont /Helvetica >>\nendobj")


# --- The rest of the file generation is unchanged ---
offsets = []
pdf_data = b"%PDF-1.4\n" 

for obj in pdf_objects:
    offsets.append(len(pdf_data))
    pdf_data += obj.encode("latin-1") + b"\n"

xref_start = len(pdf_data)
pdf_data += b"xref\n"
pdf_data += f"0 {len(pdf_objects) + 1}\n".encode("latin-1")
pdf_data += b"0000000000 65535 f \n"
for offset in offsets:
    pdf_data += f"{offset:010} 00000 n \n".encode("latin-1")

pdf_data += b"trailer\n"
pdf_data += f"<< /Size {len(pdf_objects) + 1} /Root 1 0 R >>\n".encode("latin-1")
pdf_data += b"startxref\n"
pdf_data += f"{xref_start}\n".encode("latin-1")
pdf_data += b"%%EOF\n"

with open("simple_fixed_spacing.pdf", "wb") as f:
    f.write(pdf_data)

print("PDF created: simple_fixed_spacing.pdf")
```
